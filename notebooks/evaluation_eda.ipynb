## Section 1: Load and Validate Evaluation Dataset

```python
import json
import sys
from pathlib import Path
import pandas as pd
from collections import Counter

# Add parent to path
sys.path.insert(0, str(Path.cwd().parent))

from evaluation import EvaluationDataset

# Load dataset
dataset_path = Path('../data/evaluation_dataset.json')
dataset = EvaluationDataset(str(dataset_path))
dataset.load()

print(f"Dataset loaded: {dataset.total_entries()} queries")
print(f"Dataset ID: {dataset.dataset_id}")
print(f"Category distribution: {dataset.category_distribution()}")
```

```python
# Validate dataset structure
required_fields = {
    'query_id',
    'category',
    'input_query',
    'expected_decomposition_characteristics',
    'expected_provenance_requirements',
    'expected_failure_modes',
    'expected_contradiction_behavior',
    'evaluation_notes'
}

all_valid = True
for entry in dataset.all_entries():
    missing = required_fields - set(entry.keys())
    if missing:
        print(f"Entry {entry.get('query_id')} missing fields: {missing}")
        all_valid = False

if all_valid:
    print("✓ All dataset entries valid")

# Show sample entry
sample = dataset.all_entries()[0]
print(f"\nSample entry (first query):")
for key, value in sample.items():
    if isinstance(value, (list, dict)):
        print(f"  {key}: {type(value).__name__} with {len(value)} items")
    else:
        print(f"  {key}: {value}")
```

## Section 7: Latency and Retry Distributions

```python
# Expected latency structure from test baseline
import statistics

# Simulate latency distribution for analysis structure
expected_latencies = {
    'decomposer': 80,
    'retriever': 200,
    'critic': 80,
    'synthesizer': 90,
    'orchestration': 30
}

total_expected = sum(expected_latencies.values())

print("Expected per-agent latency (ms):")
for agent, latency in expected_latencies.items():
    pct = (latency / total_expected) * 100
    print(f"  {agent:15s}: {latency:3d}ms ({pct:5.1f}%)")

print(f"\n  Total expected:  {total_expected}ms")

print("\nLatency analysis will compute:")
print("  - p50, p95, p99 percentiles")
print("  - Outliers (p99 > 2s)")
print("  - Agent contribution to variance")
print("  - Retry overhead")
```

```python
# Retry analysis structure
print("Retry pattern analysis:")
print("\nExpected retry distribution:")
print("  - ~70% of queries: 0 retries")
print("  - ~25% of queries: 1 retry")
print("  - ~4% of queries: 2 retries")
print("  - ~1% of queries: 3+ retries")

print("\nRetry overhead per iteration:")
print("  - 1st retry: +100ms")
print("  - 2nd retry: +200ms")
print("  - 3rd retry: +400ms")

print("\nRetry causes tracked:")
print("  - Retrieval timeout")
print("  - Tool failure")
print("  - Malformed response")
print("  - Transient network error")
```

## Section 8: Provenance Coverage and Contradiction Detection

```python
# Provenance analysis structure
print("Provenance Coverage Analysis:")
print("\nExpected by category:")
category_provenance = {
    'normal': 0.95,
    'ambiguous': 0.75,
    'adversarial': 0.70,
    'contradiction_prone': 0.85,
    'provenance_sensitive': 0.60
}

for category, coverage in category_provenance.items():
    print(f"  {category:20s}: {coverage:.0%} coverage expected")

print("\nProvenance gap patterns:")
print("  - Weak sources: >2 hops from authoritative source")
print("  - Unsourced claims: 0% confidence")
print("  - Ambiguous sources: Multiple conflicting attributions")
```

```python
# Contradiction detection analysis
print("Contradiction Detection Analysis:")
print("\nExpected detection rates:")
category_contradiction = {
    'normal': 0.00,  # No contradictions expected
    'ambiguous': 0.50,  # May have implicit contradictions
    'adversarial': 0.60,  # Designed to have contradictions
    'contradiction_prone': 0.95,  # Primary test for contradictions
    'provenance_sensitive': 0.50  # May have source conflicts
}

for category, rate in category_contradiction.items():
    print(f"  {category:20s}: {rate:.0%} detection expected")

print("\nContradiction types:")
print("  - Direct: A and NOT A")
print("  - Probabilistic: X>Y vs Y>X")
print("  - Temporal: X true in 2020, false in 2024")
print("  - Implicit: Y causes Z, but Z prevents Y")

print("\nMissed contradiction patterns:")
print("  - Subtle wording differences")
print("  - Domain-specific language")
print("  - Confidence-dependent statements")
```
